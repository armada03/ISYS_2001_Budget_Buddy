### Foundation Data Processing Functions

````python
# 🤖 AI Collaboration: Data Loading and Cleaning Function
# Ask AI to help you create a robust data loading function

def load_and_clean_transaction_data(file_path):
    """
    Load and clean transaction data for the Smart Finance Assistant

    🤖 AI Collaboration Prompt:
    "Create a function to load CSV transaction data with Date, Amount, Category,
    Description columns. Handle dollar signs in Amount, missing values, and
    data validation. Include clear business-focused error messages."

    Args:
        file_path: Path to CSV file or file object
    Returns:
        pandas.DataFrame: Cleaned transaction data
    """
    # Your AI-assisted implementation goes here
    # Replace this placeholder with AI-generated code
    # 🤖 AI Collaboration: Data Loading and Cleaning Function
# AI was used to help design a robust CSV cleaning function for Budget Buddy.
# I reviewed the code to make sure it handles dollar signs, missing values,
# invalid dates and required column checks.

def load_and_clean_transaction_data(file_path):
    """
    Load and clean transaction data for Budget Buddy.

    Required columns:
    - Date
    - Amount
    - Category
    - Description

    Args:
        file_path: Path to CSV file, uploaded file object, or pandas DataFrame

    Returns:
        pandas.DataFrame: Cleaned transaction data
    """

    # If the input is already a DataFrame, copy it.
    # Otherwise, read it as a CSV file.
    if isinstance(file_path, pd.DataFrame):
        df = file_path.copy()
    else:
        df = pd.read_csv(file_path)

    # Check that all required columns exist.
    required_columns = ['Date', 'Amount', 'Category', 'Description']
    missing_columns = [col for col in required_columns if col not in df.columns]

    if missing_columns:
        raise ValueError(f"Missing required columns: {missing_columns}")

    # Remove rows missing important transaction information.
    df = df.dropna(subset=['Date', 'Amount', 'Category'])

    # Clean the Amount column.
    # Example: "$45.50" becomes 45.50
    df['Amount'] = (
        df['Amount']
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .str.strip()
    )

    # Convert Amount into numeric values.
    df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

    # Remove rows where Amount could not be converted.
    df = df.dropna(subset=['Amount'])

    # Convert Date into datetime format.
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

    # Remove rows where Date could not be converted.
    df = df.dropna(subset=['Date'])

    # Clean text columns.
    df['Category'] = df['Category'].astype(str).str.strip()
    df['Description'] = df['Description'].astype(str).str.strip()

    return df


# Test the function using the uploaded transactions.csv file
clean_data = load_and_clean_transaction_data("data/transactions.csv")

print("Cleaned transaction data:")
print(clean_data.head())

print("\nData types after cleaning:")
print(clean_data.dtypes)

print("\nNumber of cleaned rows:")
print(len(clean_data))

# Test your function
# clean_data = load_and_clean_transaction_data(df_sample)
# print(clean_data)
````

````python
# 🤖 AI Collaboration: Spending Analysis Function
# AI was used to help design a spending analysis function for Budget Buddy.
# I reviewed the logic to make sure refunds are separated from normal spending
# and category percentages are calculated correctly.

def analyze_spending_patterns(df):
    """
    Analyze spending patterns and generate business insights.

    This function:
    - Separates positive spending from refunds
    - Calculates total spending
    - Calculates total refunds
    - Calculates average transaction amount
    - Summarises spending by category
    - Identifies the highest spending category

    Args:
        df: Cleaned transaction DataFrame

    Returns:
        dict: Analysis results and insights
    """

    # Separate normal spending from refunds or negative transactions.
    spending_df = df[df['Amount'] > 0].copy()
    refund_df = df[df['Amount'] < 0].copy()

    # Calculate overall spending metrics.
    total_spending = spending_df['Amount'].sum()
    total_refunds = refund_df['Amount'].sum()
    average_transaction = spending_df['Amount'].mean()
    number_of_transactions = len(spending_df)

    # Group spending by category.
    category_summary = (
        spending_df
        .groupby('Category')['Amount']
        .agg(['sum', 'mean', 'count'])
        .sort_values(by='sum', ascending=False)
    )

    # Add percentage of total spending for each category.
    if total_spending > 0:
        category_summary['percentage'] = (
            category_summary['sum'] / total_spending * 100
        ).round(2)
    else:
        category_summary['percentage'] = 0

    # Rename columns for clearer business presentation.
    category_summary = category_summary.rename(columns={
        'sum': 'Total Spent',
        'mean': 'Average Transaction',
        'count': 'Number of Transactions',
        'percentage': 'Percentage of Spending'


    })

    # Round numeric columns for cleaner presentation.
    category_summary['Total Spent'] = category_summary['Total Spent'].round(2)
    category_summary['Average Transaction'] = category_summary['Average Transaction'].round(2)
    category_summary['Percentage of Spending'] = category_summary['Percentage of Spending'].round(2)

    # Identify top spending category.
    if not category_summary.empty:
        top_category = category_summary.index[0]
        top_category_amount = category_summary.iloc[0]['Total Spent']
        top_category_percentage = category_summary.iloc[0]['Percentage of Spending']
    else:
        top_category = "None"
        top_category_amount = 0
        top_category_percentage = 0

    # Create simple business insights.
    insights = [
        f"Total positive spending was AUD {total_spending:.2f}.",
        f"Total refunds or negative transactions were AUD {total_refunds:.2f}.",
        f"The average positive transaction was AUD {average_transaction:.2f}.",
        f"The highest spending category was {top_category}, with AUD {top_category_amount:.2f} spent.",
        f"{top_category} represented {top_category_percentage:.2f}% of total positive spending."
    ]

    return {
        "total_spending": round(total_spending, 2),
        "total_refunds": round(total_refunds, 2),
        "average_transaction": round(average_transaction, 2),
        "number_of_transactions": number_of_transactions,
        "category_summary": category_summary,
        "top_category": top_category,
        "insights": insights
    }


# Test the function using the cleaned transaction data
analysis = analyze_spending_patterns(clean_data)

print("Spending summary by category:")
print(analysis["category_summary"])

print("\nKey insights:")
for insight in analysis["insights"]:
    print("-", insight)
````

````python
# 🤖 AI Collaboration: Business Insights Generator
# AI was used to help convert spending analysis results into a clear,
# user-friendly recommendation report for Budget Buddy.
# I reviewed the logic to make sure the recommendations are based on
# actual spending patterns, not generic advice.

def generate_financial_recommendations(analysis_data):
    """
    Generate actionable financial recommendations based on spending analysis.

    This function:
    - Summarises key spending results
    - Identifies the highest spending category
    - Gives practical budgeting recommendations
    - Formats the output for a non-technical user

    Args:
        analysis_data: Dictionary with spending analysis results

    Returns:
        str: Formatted recommendations report
    """

    total_spending = analysis_data["total_spending"]
    total_refunds = analysis_data["total_refunds"]
    average_transaction = analysis_data["average_transaction"]
    top_category = analysis_data["top_category"]
    category_summary = analysis_data["category_summary"]

    # Start the report with the key financial summary.
    report = f"""
BUDGET BUDDY FINANCIAL RECOMMENDATION REPORT

Overall Summary
- Total positive spending: AUD {total_spending:.2f}
- Total refunds or negative transactions: AUD {total_refunds:.2f}
- Average positive transaction: AUD {average_transaction:.2f}
- Highest spending category: {top_category}

Category Breakdown
{category_summary.to_string()}

Recommendations
"""

    # Generate advice based on the user's top spending category.
    if top_category != "None" and not category_summary.empty:
        top_category_amount = category_summary.loc[top_category, "Total Spent"]
        top_category_percentage = category_summary.loc[top_category, "Percentage of Spending"]

        report += f"""
Your highest spending category is {top_category}, with AUD {top_category_amount:.2f} spent.
This represents {top_category_percentage:.2f}% of your total positive spending.
"""

        if top_category_percentage >= 40:
            report += f"""
- {top_category} makes up a large portion of your spending.
- Consider setting a weekly or monthly spending limit for this category.
- Review whether all purchases in this category are essential.
"""
        elif top_category_percentage >= 25:
            report += f"""
- {top_category} is your largest spending area, but it is not extremely dominant.
- This category should still be monitored over multiple weeks.
- Consider comparing this category against your income or budget goal.
"""
        else:
            report += """
- Your spending appears reasonably spread across categories.
- Continue tracking your transactions to identify longer-term spending patterns.
"""
    else:
        report += """
- No positive spending categories were found.
- Check that the CSV file contains valid transaction amounts.
"""

    # Add extra advice based on average transaction size.
    if average_transaction > 50:
        report += """
- Your average transaction is above AUD 50, so reviewing larger purchases may help identify savings opportunities.
"""
    else:
        report += """
- Your average transaction is relatively moderate, so small recurring expenses may be worth reviewing.
"""

    # General budgeting advice.
    report += """
General Budgeting Advice
- Track spending weekly instead of waiting until the end of the month.
- Separate essential spending from flexible spending.
- Look for repeated small expenses, such as coffee, takeaway food or entertainment.
- Use category totals to decide where spending limits would be most useful.
"""

    return report


# Test the recommendation function using the spending analysis results
recommendations = generate_financial_recommendations(analysis)

print(recommendations)
````

## RAG System for Financial Documents

````python
# 🤖 AI Collaboration: Document Retrieval Setup
# AI was used to help design a simple RAG-style retrieval system for Budget Buddy.
# I reviewed the code to make sure it retrieves relevant budgeting guidance
# before asking the AI model to generate an answer.

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from hands_on_ai.chat import get_response

def setup_financial_rag():
    """
    Set up a simple RAG-style system for Budget Buddy.

    This system stores short financial guidance documents and retrieves
    the most relevant one based on the user's question.

    Returns:
        dict: Financial documents used for retrieval
    """

    financial_documents = {
        "budgeting_basics": """
        Budgeting means planning how your money will be spent before it is spent.
        A simple budget separates income, essential expenses, flexible spending,
        and savings. Tracking spending by category helps users identify where
        money is going and where changes can be made.
        """,

        "coffee_spending": """
        Coffee spending is a common small recurring expense. Even though one coffee
        purchase may seem small, repeated coffee purchases can become significant
        over a month. A useful strategy is setting a weekly coffee limit or making
        coffee at home several times per week.
        """,

        "grocery_spending": """
        Groceries are usually essential, but grocery spending can still be managed.
        Useful strategies include meal planning, writing a shopping list, comparing
        prices, buying only what is needed, and reducing impulse purchases.
        """,

        "entertainment_spending": """
        Entertainment is usually flexible spending. This means it can often be reduced
        more easily than essential expenses such as rent or bills. A monthly
        entertainment limit can help users enjoy leisure activities while still
        protecting their savings goals.
        """,

        "savings_goals": """
        A savings goal should have a target amount, current savings amount, and regular
        contribution amount. Dividing the remaining amount by the monthly contribution
        gives an estimate of how long it will take to reach the goal.
        """
    }

    return financial_documents


def retrieve_financial_context(user_question, financial_documents):
    """
    Retrieve the most relevant financial document based on the user's question.

    Args:
        user_question: User's finance question
        financial_documents: Dictionary of financial guidance documents

    Returns:
        str: Most relevant document text
    """

    document_titles = list(financial_documents.keys())
    document_texts = list(financial_documents.values())

    # Add the user question to the list so it can be compared with the documents.
    all_texts = document_texts + [user_question]

    vectorizer = TfidfVectorizer(stop_words="english")
    tfidf_matrix = vectorizer.fit_transform(all_texts)

    # Compare the user question with all documents.
    question_vector = tfidf_matrix[-1]
    document_vectors = tfidf_matrix[:-1]

    similarity_scores = cosine_similarity(question_vector, document_vectors).flatten()

    best_match_index = similarity_scores.argmax()
    best_title = document_titles[best_match_index]
    best_context = document_texts[best_match_index]

    return f"Retrieved topic: {best_title}\n\n{best_context}"


def rag_finance_answer(user_question):
    """
    Answer a finance question using retrieved context and hands-on-ai.

    Args:
        user_question: User's finance question

    Returns:
        str: RAG-style finance answer
    """

    financial_documents = setup_financial_rag()
    retrieved_context = retrieve_financial_context(user_question, financial_documents)

    prompt = f"""
    You are Budget Buddy, a friendly personal finance assistant.

    Use the retrieved context below to answer the user's question.
    Keep the answer practical, simple and student-friendly.

    Retrieved context:
    {retrieved_context}

    User question:
    {user_question}
    """

    try:
        response = get_response(prompt)

        if "Error" in str(response) or "Connection error" in str(response):
            return f"""
The AI server was unavailable, but Budget Buddy retrieved this relevant guidance:

{retrieved_context}
"""

        return response

    except Exception:
        return f"""
The AI server was unavailable, but Budget Buddy retrieved this relevant guidance:

{retrieved_context}
"""


# Test the RAG system
rag_answer = rag_finance_answer(
    "How to budget properly?"
)

print(rag_answer)
````

## Custom Financial Tools

````python
# 🤖 AI Collaboration: Custom Tool Development
# AI was used to help design a savings goal calculator for Budget Buddy.
# I reviewed the logic to make sure it handles invalid inputs, reached goals,
# and produces a user-friendly result.

def savings_goal_calculator(current_savings, monthly_contribution, target_amount):
    """
    Calculate how long it will take a user to reach a savings goal.

    Args:
        current_savings: Amount the user has already saved
        monthly_contribution: Amount the user plans to save each month
        target_amount: Savings goal amount

    Returns:
        str: Formatted savings goal result
    """

    try:
        current_savings = float(current_savings)
        monthly_contribution = float(monthly_contribution)
        target_amount = float(target_amount)
    except ValueError:
        return "Please enter valid numbers for current savings, monthly contribution and target amount."

    if current_savings < 0 or monthly_contribution < 0 or target_amount < 0:
        return "Savings values cannot be negative."

    if target_amount <= current_savings:
        return (
            f"You have already reached your savings goal. "
            f"Current savings: AUD {current_savings:.2f}, "
            f"Target: AUD {target_amount:.2f}."
        )

    if monthly_contribution == 0:
        return "Monthly contribution must be greater than zero to calculate a savings timeline."

    remaining_amount = target_amount - current_savings
    months_needed = remaining_amount / monthly_contribution

    return f"""
SAVINGS GOAL CALCULATOR

Current savings: AUD {current_savings:.2f}
Target amount: AUD {target_amount:.2f}
Monthly contribution: AUD {monthly_contribution:.2f}

Remaining amount needed: AUD {remaining_amount:.2f}
Estimated time to reach goal: {months_needed:.1f} months

Recommendation:
To reach your goal faster, consider increasing your monthly contribution or reducing spending in flexible categories such as entertainment, dining or coffee.
"""


def create_savings_calculator_tool():
    """
    Create the savings goal calculator tool for Budget Buddy.

    Returns:
        function: savings_goal_calculator function
    """

    return savings_goal_calculator


# Create the tool
savings_tool = create_savings_calculator_tool()

# Test the tool
savings_result = savings_tool(
    current_savings=700,
    monthly_contribution=300,
    target_amount=5000
)

print(savings_result)
````

````python
import gradio as gr
import gradio.blocks

def analyse_uploaded_csv(file):
    """
    Analyse an uploaded transaction CSV file and return a recommendation report.
    """

    try:
        if file is None:
            return "Please upload a CSV file first."

        cleaned_data = load_and_clean_transaction_data(file.name)
        analysis_results = analyze_spending_patterns(cleaned_data)
        recommendation_report = generate_financial_recommendations(analysis_results)

        return recommendation_report

    except Exception as e:
        return f"Error analysing CSV file: {e}"


def chatbot_interface(user_question):
    """
    Connect the Gradio chatbot tab to the Budget Buddy chatbot function.
    """

    if not user_question or user_question.strip() == "":
        return "Please enter a finance question."

    try:
        return budget_buddy_chat(user_question)
    except Exception as e:
        return f"Chatbot error: {e}"


def rag_interface(user_question):
    """
    Connect the Gradio RAG tab to the Budget Buddy RAG function.
    """

    if not user_question or user_question.strip() == "":
        return "Please enter a question for the RAG system."

    try:
        return rag_finance_answer(user_question)
    except Exception as e:
        return f"RAG system error: {e}"


def savings_calculator_interface(current_savings, monthly_contribution, target_amount):
    """
    Connect the Gradio savings tab to the savings goal calculator.
    """

    try:
        return savings_goal_calculator(current_savings, monthly_contribution, target_amount)
    except Exception as e:
        return f"Savings calculator error: {e}"


def create_finance_assistant_ui():
    """
    Create a Gradio interface for Budget Buddy.

    The interface includes:
    - CSV Spending Analysis
    - Finance Chatbot
    - RAG Financial Guidance
    - Savings Goal Calculator

    Returns:
        gr.Blocks: Gradio application
    """

    with gr.Blocks(title="Budget Buddy - Smart Finance Assistant") as demo:
        gr.Markdown("# Budget Buddy - Smart Finance Assistant")
        gr.Markdown(
            "Upload transaction data, analyse spending, ask finance questions, "
            "retrieve budgeting guidance, and calculate savings goals."
        )

        with gr.Tab("CSV Spending Analysis"):
            gr.Markdown("### Upload a transaction CSV file")
            gr.Markdown("Required columns: Date, Amount, Category, Description")

            csv_file = gr.File(label="Upload transactions.csv")
            csv_output = gr.Textbox(
                label="Budget Buddy Spending Report",
                lines=20
            )

            analyse_button = gr.Button("Analyse Spending")
            analyse_button.click(
                fn=analyse_uploaded_csv,
                inputs=csv_file,
                outputs=csv_output
            )

        with gr.Tab("Finance Chatbot"):
            gr.Markdown("### Ask Budget Buddy a finance question")

            chat_input = gr.Textbox(
                label="Your question",
                placeholder="Example: I spend too much on coffee, what should I do?"
            )

            chat_output = gr.Textbox(
                label="Budget Buddy Response",
                lines=10
            )

            chat_button = gr.Button("Ask Budget Buddy")
            chat_button.click(
                fn=chatbot_interface,
                inputs=chat_input,
                outputs=chat_output
            )

        with gr.Tab("RAG Financial Guidance"):
            gr.Markdown("### Ask a question using retrieved budgeting guidance")

            rag_input = gr.Textbox(
                label="Your RAG question",
                placeholder="Example: What's a good strategy for overspending on entertainment?"
            )

            rag_output = gr.Textbox(
                label="RAG Response",
                lines=12
            )

            rag_button = gr.Button("Get RAG Guidance")
            rag_button.click(
                fn=rag_interface,
                inputs=rag_input,
                outputs=rag_output
            )

        with gr.Tab("Savings Goal Calculator"):
            gr.Markdown("### Calculate how long it will take to reach a savings goal")

            current_savings = gr.Number(label="Current Savings AUD", value=500)
            monthly_contribution = gr.Number(label="Monthly Contribution AUD", value=250)
            target_amount = gr.Number(label="Target Amount AUD", value=2000)

            savings_output = gr.Textbox(
                label="Savings Goal Result",
                lines=10
            )

            savings_button = gr.Button("Calculate Savings Goal")
            savings_button.click(
                fn=savings_calculator_interface,
                inputs=[current_savings, monthly_contribution, target_amount],
                outputs=savings_output
            )

    return demo


# Launch the complete Budget Buddy application
demo = create_finance_assistant_ui()
demo.launch()
````
